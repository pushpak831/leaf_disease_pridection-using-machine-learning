# Plant Disease Prediction: Test notebook

Tests the trained model on a single leaf image, step by step.
Keep this notebook in the project folder (next to `main.py`) and open that folder in VS Code.

# Importing Libraries

```python
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

from plant_labels import CLASS_NAMES, IMG_SIZE, load_model, pretty
```

Dataset Link: https://www.kaggle.com/datasets/vipoooool/new-plant-diseases-dataset

The 38 class names (in the model's order) come from `plant_labels.py`, so the dataset
folders are **not** needed to run this notebook.

# Class names

```python
class_name = CLASS_NAMES
print(len(class_name), "classes")
print(class_name)
```

# Loading Model

```python
cnn = load_model()          # looks for trained_plant_disease_model.keras next to this notebook
cnn.summary()
```

# Visualising and Performing Prediction on Single image

```python
# Change this to any leaf photo (a dataset test image, or sample_images/rust_leaf.jpg)
image_path = 'sample_images/rust_leaf.jpg'

image = tf.keras.preprocessing.image.load_img(image_path)        # original size, for display
plt.imshow(image)
plt.title('Test Image')
plt.xticks([])
plt.yticks([])
plt.show()
```

## Testing Model

```python
small = tf.keras.preprocessing.image.load_img(image_path, target_size=(IMG_SIZE, IMG_SIZE), interpolation='bilinear')
input_arr = tf.keras.preprocessing.image.img_to_array(small)
input_arr = np.array([input_arr])  # Convert single image to a batch.
predictions = cnn.predict(input_arr)
```

```python
print(predictions)
```

```python
result_index = np.argmax(predictions)  # Return index of max element
print(result_index)
```

# Displaying the disease prediction

```python
model_prediction = class_name[result_index]
plant, disease = pretty(model_prediction)
confidence = predictions[0][result_index] * 100

plt.imshow(image)
plt.title(f"Disease Name: {model_prediction}  ({confidence:.1f}%)")
plt.xticks([])
plt.yticks([])
plt.show()
```

## Top 3 possibilities

```python
top3 = np.argsort(predictions[0])[::-1][:3]
for i in top3:
    p, d = pretty(class_name[i])
    print(f"{p:28s} {d or 'healthy':45s} {predictions[0][i] * 100:5.1f}%")
```
